```python
import pandas as pd

def get_formatted_table(df, target_l0):
    # 1. Find the average L0 for each variant (hyperparameter config) across its 3 seeds
    variant_l0 = df.groupby(['model_class', 'variant_name'])['l0_active_features'].mean().reset_index()
    
    # 2. For each architecture, find the variant closest to the target L0
    best_idx = variant_l0.groupby('model_class')['l0_active_features'].apply(
        lambda x: (x - target_l0).abs().idxmin()
    )
    best_variants = variant_l0.loc[best_idx, ['model_class', 'variant_name']]
    
    # 3. Filter the original dataframe to ONLY include the 3 seeds of those best variants
    best_runs = df.merge(best_variants, on=['model_class', 'variant_name'])
    
    # 4. Helper function to format the numbers as "mean [min, max]"
    def format_stat(x, is_pct=False, decimals=3):
        val_center = x.mean() # Change to x.median() here if you want median instead
        val_min = x.min()
        val_max = x.max()
        
        if is_pct:
            return f"{val_center:.1f}% [{val_min:.1f}, {val_max:.1f}]"
        else:
            return f"{val_center:.{decimals}f} [{val_min:.{decimals}f}, {val_max:.{decimals}f}]"

    # 5. Group by model_class and apply the formatting
    records = []
    for model_class, group in best_runs.groupby('model_class'):
        records.append({
            'Name' : group['Name'].iloc[0],
            'Target L0': f"~ {target_l0}",
            'Architecture': model_class,
            'Achieved L0': format_stat(group['l0_active_features'], decimals=1),
            'Dead Neurons.': format_stat(group['dead_neurons_pct_window'], is_pct=True),
            'Expl. Var.': format_stat(group['explained_variance'], decimals=3),
            'p50 Coh.': format_stat(group['dictionary_coherence_nn/p50'], decimals=3),
            'p90 Coh.': format_stat(group['dictionary_coherence_nn/p90'], decimals=3),
            'Max Coh.': format_stat(group['dictionary_coherence_abs_max'], decimals=3),
            'Shrinkage': format_stat(group['feature_shrinkage_ratio'], decimals=3),
            'Norm. Ent.': format_stat(group['feature_utilization_batch/normalized_entropy'], decimals=3),
            'Top 10 mass;': format_stat(group['feature_utilization_batch/top10pct_mass_pct'], decimals=3),
            'Val final patched loss': format_stat(group['val_final_patched_loss'], decimals=3),
            'Val final KL div.': format_stat(group['val_final_kl_div/mean'], decimals=3),
        })

        
    return pd.DataFrame(records)


table_16 = get_formatted_table(df, target_l0=16)
table_32 = get_formatted_table(df, target_l0=32)
table_128 = get_formatted_table(df, target_l0=128)

print("Table for Target L0 = 16")
print(table_16.to_string(index=False))

print("\nTable for Target L0 = 32")
print(table_32.to_string(index=False))

print("\nTable for Target L0 = 128")
print(table_128.to_string(index=False))
```
